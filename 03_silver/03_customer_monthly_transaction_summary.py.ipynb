Business scenario

The bank wants a monthly transaction summary for each customer.
calculate, for each:

- customer_id
- transaction month

the following:

- Total transaction amount
- Number of transactions
- Average transaction amount
- Maximum transaction amount

```sql
%python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

BRONZE_PATH = "/mnt/usvikformula1dl/bronze"
SILVER_PATH = "/mnt/usvikformula1dl/silver"


transaction_bronze_path = f"{BRONZE_PATH}/transaction_fact"
transaction_df=spark.read.format("delta").load(f"{transaction_bronze_path}")
customer_monthly_summary=transaction_df.groupBy("customer_id",F.date_format("transaction_ts", "yyyy-MM").alias("transaction_month")).agg(F.count("transaction_ts").alias("monthly_tran_ct"),F.sum("amount").alias("monthly_tran_amt"),F.avg("amount").alias("monthly_tran_avg"),F.max("amount").alias("_max_amount"))

display(customer_monthly_summary)


#cus_monthly_summary_df=transaction_df.
```
