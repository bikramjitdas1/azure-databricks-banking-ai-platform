The bank wants to identify the most recent transaction for every account.


```sql
%python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

BRONZE_PATH = "/mnt/usvikformula1dl/bronze"
transaction_bronze_path = f"{BRONZE_PATH}/transaction_fact"
transaction_df=spark.read.format("delta").load(f"{transaction_bronze_path}")

window_spec=Window.partitionBy("account_id").orderBy(F.col("transaction_ts").desc(),F.col("transaction_id").desc())

recent_tran_df=transaction_df.withColumn("row_num",F.row_number().over(window_spec)).filter("row_num==1").drop("row_num")
display(recent_tran_df)


```
