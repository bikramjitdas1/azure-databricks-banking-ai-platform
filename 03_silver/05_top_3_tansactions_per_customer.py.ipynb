The bank wants to identify the top 3 highest-value transactions for each customer.

```sql
%python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

BRONZE_PATH = "/mnt/usvikformula1dl/bronze"
transaction_bronze_path = f"{BRONZE_PATH}/transaction_fact"
transaction_df=spark.read.format("delta").load(f"{transaction_bronze_path}")

window_spec=Window().partitionBy("customer_id").orderBy(F.col("amount").desc())
top_3_tran_df=transaction_df.withColumn("row_num",F.rank().over(window_spec)).filter(F.col("row_num") <= 3)
display(top_3_tran_df)
```
