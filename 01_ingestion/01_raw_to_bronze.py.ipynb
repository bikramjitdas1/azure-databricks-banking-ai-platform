```sql
%python
from pyspark.sql import functions as f
RAW_PATH = "/mnt/usvikformula1dl/raw"
BRONZE_PATH = "/mnt/usvikformula1dl/bronze"

customer_raw_path = f"{RAW_PATH}/customer_master"
customer_bronze_path = f"{BRONZE_PATH}/customer_master"

customer_master_df = (spark.read.format("json")
.option("inferschema", True)
.load(customer_raw_path))

customer_master_df_trans=customer_master_df.withColumn("ingestion_timestamp",f.current_timestamp()).withColumn("source_file", f.col("_metadata.file_path"))
customer_master_df_trans.write.mode("overwrite").format("delta").save(customer_bronze_path)



#verification :
bronze_df=spark.read.format("delta").load(customer_bronze_path)
bronze_df.display()
```
